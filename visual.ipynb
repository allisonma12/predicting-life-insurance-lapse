```python
# import libraries, these are alr downloaded from beforehand 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv('data/CleanedLifeInsurance.csv')
df.head()
```

```python
df['PAYMENT MODE'].value_counts().plot(
    kind='bar', 
    color='lightblue',
    rot = 45
    )
```

```python
df['POLICY STATUS'].value_counts().plot(
    kind='bar', 
    color='lightblue',
    rot = 45
    )
```

```python
# create a crosstab and normalize by row (index) to get percentages
crosstab_pct = pd.crosstab(df['PAYMENT MODE'], df['POLICY STATUS'], normalize='index') * 100
crosstab_pct
```

```python
# plot as a stacked bar chart
crosstab_pct.plot(
    kind='bar', 
    stacked=True, 
    figsize=(10, 6), 
    colormap='mako'
    )

# titles and labels
plt.title('Policy Status Breakdown by Payment Mode (%)', fontsize=14, fontweight='bold')
plt.xlabel('Payment Mode', fontsize=12)
plt.ylabel('Percentage (%)', fontsize=12)
plt.xticks(rotation = 0)
plt.legend(title='Policy Status', bbox_to_anchor=(1.05, 1), loc='upper left')

# Show the plot     
plt.show()
```
